```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
import os

print("Dataset 1:")
print(os.listdir('/content/archive (1)'))

print("\nDataset 2:")
print(os.listdir('/content/archive (2)'))
```

```python
english_df = pd.read_csv('/content/archive (1)/sentiment_data.csv')
arabic_df = pd.read_csv('/content/archive (2)/arabic_sentiment_reviews.csv')

print("English Dataset:")
print(english_df.head())

print("\nArabic Dataset:")
print(arabic_df.head())
```

```python
print("English labels:")
print(english_df['Sentiment'].value_counts().sort_index())

print("\nArabic labels:")
print(arabic_df['label'].value_counts().sort_index())
```

```python
# تجهيز البيانات الإنجليزية
english_clean = english_df[['Comment', 'Sentiment']].copy()

english_clean.columns = ['text', 'sentiment']

english_clean['language'] = 'English'

english_clean['sentiment'] = english_clean['sentiment'].map({
    0: 'Negative',
    1: 'Neutral',
    2: 'Positive'
})


# تجهيز البيانات العربية
arabic_clean = arabic_df[['content', 'label']].copy()

arabic_clean.columns = ['text', 'sentiment']

arabic_clean['language'] = 'Arabic'

arabic_clean['sentiment'] = arabic_clean['sentiment'].map({
    0: 'Negative',
    1: 'Positive'
})


print("English:")
print(english_clean.head())

print("\nArabic:")
print(arabic_clean.head())
```

```python
# دمج البيانات العربية والإنجليزية
df = pd.concat([english_clean, arabic_clean], ignore_index=True)

# حذف التعليقات الفارغة
df = df.dropna(subset=['text', 'sentiment'])

# تحويل النص إلى String
df['text'] = df['text'].astype(str)

# حذف التعليقات المكررة
df = df.drop_duplicates(subset=['text'])

# إعادة ترتيب الصفوف
df = df.reset_index(drop=True)

print("Number of comments:", len(df))
print("\nSentiment distribution:")
print(df['sentiment'].value_counts())

print("\nLanguage distribution:")
print(df['language'].value_counts())
```

```python
import re

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+', '', text)  # حذف الروابط
    text = re.sub(r'<.*?>', '', text)            # حذف HTML
    text = re.sub(r'\s+', ' ', text).strip()     # ترتيب المسافات
    return text

df['clean_text'] = df['text'].apply(clean_text)

print(df[['text', 'clean_text']].head(10))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['clean_text'],
    df['sentiment'],
    test_size=0.2,
    random_state=42,
    stratify=df['sentiment']
)

print("Training data:", len(X_train))
print("Testing data:", len(X_test))
```

```python
tfidf = TfidfVectorizer(
    max_features=10000,
    ngram_range=(1, 2)
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Training shape:", X_train_tfidf.shape)
print("Testing shape:", X_test_tfidf.shape)
```

```python
model = LogisticRegression(
    max_iter=1000
)

model.fit(X_train_tfidf, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test_tfidf)

print(y_pred[:20])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.2%}")
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    xticklabels=model.classes_,
    yticklabels=model.classes_
)

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Sentiment Analysis Confusion Matrix')
plt.show()
```

```python
new_comments = [
    "I absolutely love this product, it is amazing!",
    "This is the worst experience ever.",
    "The product is okay, nothing special.",
    "I really enjoyed using this app.",
    "I am disappointed with the service."
]

new_comments_tfidf = tfidf.transform(new_comments)

predictions = model.predict(new_comments_tfidf)

for comment, prediction in zip(new_comments, predictions):
    print(f"Comment: {comment}")
    print(f"Sentiment: {prediction}")
    print("-" * 50)
```

```python
arabic_comments = [
    "الخدمة ممتازة وأنصح الجميع بها",
    "التجربة سيئة جدًا ولم تعجبني",
    "المنتج عادي وليس سيئًا",
    "أنا سعيد جدًا بهذه الخدمة",
    "تجربة سيئة ولن أكررها"
]

arabic_comments_tfidf = tfidf.transform(arabic_comments)

arabic_predictions = model.predict(arabic_comments_tfidf)

for comment, prediction in zip(arabic_comments, arabic_predictions):
    print(f"Comment: {comment}")
    print(f"Sentiment: {prediction}")
    print("-" * 50)
```
